### 3. Wordcloud - Từ khóa

**Mô tả:** Biểu đồ Wordcloud trích xuất từ văn bản bình luận của khách hàng.

**Kết luận:** Các từ khóa nổi bật nhất là **"great", "clean", "location", "host", "nice"**. Yếu tố được quan tâm hàng đầu là Vị trí (location) và Sự sạch sẽ (clean). Thái độ nhiệt tình của chủ nhà (host) cũng là một điểm cộng lớn khiến khách hàng sẵn sàng giới thiệu (recommend).

```r
listings <- read_csv('../../data/cleaned_listings.csv', show_col_types = FALSE)

ggplot(listings, aes(x = price)) +
  geom_histogram(fill = "skyblue", color = "black", bins = 50) +
  coord_cartesian(xlim = c(0, 1000)) +
  theme_minimal() +
  labs(title = "Phân bố giá phòng Airbnb tại NYC (Dưới $1000)", x = "Giá (USD)", y = "Số lượng phòng")
```
