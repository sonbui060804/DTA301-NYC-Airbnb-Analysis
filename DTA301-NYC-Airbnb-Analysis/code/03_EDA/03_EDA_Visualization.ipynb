# PHASE 3: Exploratory Data Analysis (EDA) & Visualization
This notebook loads the cleaned datasets and generates visualizations for the NYC Airbnb project.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Cài đặt giao diện
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

### 1. Phân bố giá phòng (Price Distribution)

```
# Đọc dữ liệu
listings = pd.read_csv('../../data/cleaned_listings.csv')

# Vẽ biểu đồ phân bố giá
plt.figure(figsize=(10, 6))
sns.histplot(listings['price'], bins=50, kde=True, color='skyblue')
plt.title('Phân bố giá phòng Airbnb tại NYC (Dưới $2000)', fontsize=14)
plt.xlabel('Giá (USD)')
plt.ylabel('Số lượng phòng')
plt.xlim(0, 1000) # Phóng to vùng dữ liệu chính
plt.show()
```

### 2. Bản đồ mật độ Airbnb tại New York

```
plt.figure(figsize=(12, 10))
sns.scatterplot(data=listings, x='longitude', y='latitude', hue='neighbourhood_group', 
                palette='Set2', s=10, alpha=0.6)
plt.title('Bản đồ Vị trí các phòng Airbnb tại NYC', fontsize=16)
plt.xlabel('Kinh độ (Longitude)')
plt.ylabel('Vĩ độ (Latitude)')
plt.legend(title='Quận (Neighbourhood Group)', markerscale=3)
plt.show()
```

### 3. Wordcloud - Từ khóa khách hàng hay dùng nhất (Reviews)

```
# Cần cài thư viện wordcloud: pip install wordcloud
from wordcloud import WordCloud

# Đọc file nén reviews xz
reviews = pd.read_csv('../../data/cleaned_reviews.csv.xz', compression='xz')

# Lấy mẫu 50,000 dòng để vẽ nhanh
sample_reviews = reviews['comments'].dropna().sample(50000, random_state=42)
text = ' '.join(sample_reviews.astype(str).values)

wordcloud = WordCloud(width=800, height=400, background_color='white', colormap='viridis', max_words=100).generate(text)

plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Wordcloud - Đánh giá của khách hàng về Airbnb NYC', fontsize=16)
plt.show()
```
