# PHASE 3: Exploratory Data Analysis (EDA) & Visualization
This notebook loads the cleaned datasets and generates visualizations for the NYC Airbnb project using R.

```r
# Cài đặt các gói nếu chưa có
# install.packages(c("ggplot2", "dplyr", "readr", "wordcloud", "tm"))

library(ggplot2)
library(dplyr)
library(readr)
```

### 1. Phân bố giá phòng (Price Distribution)

```r
# Đọc dữ liệu
listings <- read_csv('../../data/cleaned_listings.csv', show_col_types = FALSE)

# Vẽ biểu đồ phân bố giá
ggplot(listings, aes(x = price)) +
  geom_histogram(fill = "skyblue", color = "black", bins = 50) +
  coord_cartesian(xlim = c(0, 1000)) +
  theme_minimal() +
  labs(title = "Phân bố giá phòng Airbnb tại NYC (Dưới $1000)",
       x = "Giá (USD)", y = "Số lượng phòng")
```

### 2. Bản đồ mật độ Airbnb tại New York

```r
ggplot(listings, aes(x = longitude, y = latitude, color = neighbourhood_group)) +
  geom_point(alpha = 0.6, size = 0.5) +
  theme_minimal() +
  guides(color = guide_legend(override.aes = list(size = 3))) +
  labs(title = "Bản đồ Vị trí các phòng Airbnb tại NYC",
       x = "Kinh độ", y = "Vĩ độ", color = "Quận")
```

### 3. Wordcloud - Từ khóa khách hàng hay dùng nhất (Reviews)

```r
library(wordcloud)
library(tm)

# Đọc file nén reviews xz (Cần gói archive hoặc readr phiên bản mới)
reviews <- read_csv('../../data/cleaned_reviews.csv.xz', show_col_types = FALSE)

# Lấy ngẫu nhiên một mẫu nhỏ để vẽ cho nhanh
set.seed(42)
sample_reviews <- reviews %>% 
  filter(!is.na(comments)) %>% 
  sample_n(20000)

# Xử lý text
corpus <- Corpus(VectorSource(sample_reviews$comments))
corpus <- tm_map(corpus, content_transformer(tolower))
corpus <- tm_map(corpus, removePunctuation)
corpus <- tm_map(corpus, removeWords, stopwords("english"))

# Vẽ wordcloud
wordcloud(corpus, max.words = 100, random.order = FALSE, colors = brewer.pal(8, "Dark2"))
```
